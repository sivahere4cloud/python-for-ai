```python
s = "python"

s[0]      # 'p'
s[-1]     # 'n'    negative index counts from the end
s[1:4]    # 'yth'  start included, stop excluded
s[:3]     # 'pyt'
s[3:]     # 'hon'
s[::2]    # 'pto'  step of 2
s[::-1]   # 'nohtyp'  reversed
len(s)    # 6
```

```python
name = "Siva"
price = 3.14159

print(f"{name!r}")       # 'Siva'     (repr)
print(f"{price:.2f}")    # 3.14       (2 decimal places)
print(f"{42:>6}")        #     42     (right-aligned, width 6)
print(f"{1_000_000:,}")  # 1,000,000
print(f"{price=}")       # price=3.14159   (debug form)
```
